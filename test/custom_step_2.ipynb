```python
import pandas as pd

df_predict = pd.read_csv("/mnt/data/image_recognition/brown_forman_req/puru_output/predictions_full.csv")
df_brown_forman = pd.read_csv("/mnt/data/image_recognition/brown_forman_req/puru_output/Brown Forman Universe Data - precision and recall.csv")
```

```python
print(df_predict.columns)
```

```python
df_brown_forman = df_brown_forman.loc[
    :, ~df_brown_forman.columns.str.contains('^Unnamed')
]
df_brown_forman.rename(columns={"Sales Team Prediction":"Category"}, inplace=True)
print(df_brown_forman.columns)
```

```python
df_predict.drop(columns = ['Category'], inplace=True)
df_brown_forman_specific = df_brown_forman[['address','Category']]

print(df_predict.columns)
print(df_brown_forman_specific.columns)
```

```python
df = pd.merge(df_predict, df_brown_forman_specific , on='address', how='left')

print(df.columns)
```

```python
import numpy as np
df.loc[df['Category'] == 'Influencer', 'Category'] = 'Super Premium'
df.loc[df['Category'] == 'Thane', 'Category'] = np.nan
df = df[df['Category'] != 'Beer licence']
print(df['Category'].value_counts())
df.shape
```

```python
df.to_csv("/mnt/data/image_recognition/brown_forman_req/puru_input/training_data_bf_22_may.csv")
```

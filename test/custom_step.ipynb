```python
import pandas as pd
from sklearn.metrics import classification_report

# Read file
df = pd.read_csv("/mnt/data/image_recognition/brown_forman_req/puru_output/predictions_full.csv")

# Rename columns if needed
# Actual category = BF categorization (Column E)
# Predicted category = prediction column

actual_col = "Category"    # replace with exact column name if different
pred_col = "prediction"

# Remove rows where actual category is Thane
df_filtered = df[df[actual_col].str.strip().ne("Thane")]

# Remove missing values
df_filtered = df_filtered.dropna(subset=[actual_col, pred_col])

# Precision, Recall, F1 for each category
report = classification_report(
    df_filtered[actual_col],
    df_filtered[pred_col],
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()

# Keep only category rows
metrics_df = report_df[
    ['precision', 'recall', 'f1-score', 'support']
]

print(metrics_df)

# # Optional: save to Excel
metrics_df.to_csv(os.path.join(OUTPUT_DIR, "metric_precision_recall.csv"))
```

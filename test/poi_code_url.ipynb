```python
import pandas as pd
data = pd.read_csv("//mnt/data/image_recognition/brown_forman_req/input/brown_forman_poi_data.csv")
data.columns
```

```python
data.rename(columns={"poi_code_primary": "poi_code"}, inplace=True)
data.columns
```

```python
data_columns = pd.read_csv("/mnt/data/image_recognition/brown_forman_req/output/restaurant_brand_data_columns.csv")
data_columns.columns
```

```python
data_all = pd.read_csv("/mnt/data/image_recognition/brown_forman_req/output/restaurant_brand_data_include_all.csv")
data_all.columns
```

```python

# Inner merge
merged_inner = pd.merge(data_columns, data, 
                        on='poi_code', 
                        how='left')

merged_inner = merged_inner.loc[:, ~merged_inner.columns.duplicated()]
merged_inner.columns
```

```python

# Outer merge
final_df = pd.merge(data, data_all, 
                    on='poi_code', 
                    how='outer')

final_df = final_df.loc[:, ~final_df.columns.duplicated()]
final_df.columns
```

```python
merged_inner.to_csv("/mnt/data/image_recognition/brown_forman_req/output/restaurant_brand_data_columns_attributes.csv", index=False)
```

```python
final_df.to_csv("/mnt/data/image_recognition/brown_forman_req/output/restaurant_brand_data_include_all_attributes.csv", index=False)
```

```python
import pandas as pd
mentions = pd.read_csv("/mnt/data/image_recognition/brown_forman_req/input/browformanbars.csv")

mentions.columns
mentions.rename(columns={" source": "mentions", "hotels":"restaurant_name"}, inplace=True)
mentions.columns
```

```python
brand_data = pd.read_csv("/mnt/data/image_recognition/brown_forman_req/output/restaurant_brand_data_columns_attributes.csv")
brand_data.columns
```

```python
brand_data_all = pd.read_csv("/mnt/data/image_recognition/brown_forman_req/output/restaurant_brand_data_include_all_attributes.csv")
brand_data_all.columns
```

```python
data_all = pd.merge(
    brand_data_all,
    mentions,
    on = "poi_code",
    how="left"
)

data_brand = pd.merge(
    brand_data,
    mentions,
    on = "poi_code",
    how="left"
)
```

```python
data_all.to_csv("/mnt/data/image_recognition/brown_forman_req/output/restaurant_brand_data_include_all_attributes_mentions.csv", index=False)
data_brand.to_csv("/mnt/data/image_recognition/brown_forman_req/output/restaurant_brand_data_columns_attributes_mentions.csv", index=False)
```

```python
file_name_address = pd.read_csv("/mnt/data/image_recognition/brown_forman_req/input/uploaded_data_1765.csv")
file_name_address.columns
```

```python
df = data_all.merge(
    file_name_address[['filename', 'name', 'address']],
    on='filename',
    how='left'
)

# restaurant_name
if 'restaurant_name' in df.columns:
    df['restaurant_name'] = df['restaurant_name'].fillna(df['name'])
else:
    df['restaurant_name'] = df['name']

# address handling (safe)
if 'address_x' in df.columns:
    df['address'] = df['address_x'].fillna(df['address_y'])
    df = df.drop(columns=['address_x', 'address_y'])
else:
    df['address'] = df['address']  # already from file_name_address

# drop helper
df = df.drop(columns=['name'])
```

```python
df.to_csv("/mnt/data/image_recognition/brown_forman_req/output/restaurant_brand_data_include_all_attributes_mentions.csv", index=False)
```
